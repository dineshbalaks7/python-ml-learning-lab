# Pandas Learning Guide and Practice

A practical guide to the pandas functions and workflows used most often in data analysis. Run the next cell first; the examples use their own `sales` DataFrame so they do not depend on earlier notebook variables.

**Topics:** Series and DataFrames, inspecting data, selecting and filtering, sorting, creating columns, missing values, duplicates, summaries, grouping, combining tables, text and dates, reshaping, and CSV files.

For more detail on any method, use `help(pd.DataFrame.method_name)` or `pd.DataFrame.method_name?` in Jupyter.

```python
import pandas as pd

# A small, reproducible dataset for the examples and exercises.
sales = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006, 1006],
    "date": ["2025-01-03", "2025-01-05", "2025-02-10", "2025-02-12", "2025-03-01", "2025-03-04", "2025-03-04"],
    "customer": ["Asha", "Ben", "Asha", "Chen", "Ben", "Dia", "Dia"],
    "city": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune", "Pune"],
    "product": ["Notebook", "Pen", "Pen", "Notebook", "Pencil", "Notebook", "Notebook"],
    "units": [3, 10, 5, 2, 8, 4, 4],
    "unit_price": [120.0, 15.0, 15.0, 120.0, 8.0, 120.0, 120.0],
    "discount": [0.0, 0.10, 0.0, 0.05, None, 0.10, 0.10]
})
sales["revenue"] = sales["units"] * sales["unit_price"] * (1 - sales["discount"].fillna(0))
sales
```

## 1. Core objects and inspection

- `pd.Series(values, index=...)` creates a labeled one-dimensional array; `pd.DataFrame(data)` creates a table.
- Inspect with `head()`, `tail()`, `sample()`, `shape`, `columns`, `index`, `dtypes`, `info()`, and `describe()`.
- `value_counts()` counts category values; `unique()` and `nunique()` list/count distinct values.

## 2. Select, filter, and sort

- Select a column with `sales["city"]` (Series) or several with `sales[["city", "revenue"]]` (DataFrame).
- `loc` selects by labels/conditions; `iloc` selects by integer positions. Use `at` and `iat` for one value.
- Filter with boolean masks, `isin()`, `between()`, or `query()`; combine masks with `&`, `|`, and `~` (parenthesize each condition).
- Order rows with `sort_values()` or `sort_index()`.

## 3. Clean and transform

- Add or transform columns with vectorized arithmetic, `assign()`, `map()`, `replace()`, and `apply()`.
- Change types with `astype()`; rename labels with `rename()`.
- Find missing values with `isna()` / `notna()`, remove rows with `dropna()`, or fill with `fillna()`.
- Detect/remove duplicates with `duplicated()` and `drop_duplicates()`.

## 4. Summarize and combine

- Numeric summaries include `sum()`, `mean()`, `median()`, `min()`, `max()`, and `std()`; use `agg()` to request several at once.
- `groupby(...).agg(...)` computes summaries per group; `transform()` returns group results aligned to original rows.
- `pd.concat()` stacks tables; `merge()` joins tables by key (`how="inner"`, `"left"`, `"right"`, or `"outer"`).

## 5. Text, dates, reshape, and files

- String operations use the `.str` accessor, e.g. `str.lower()`, `str.strip()`, `str.contains()`, and `str.replace()`.
- Parse dates with `pd.to_datetime()` and use `.dt.year`, `.dt.month`, `.dt.day`, or `.dt.day_name()`.
- `pivot_table()` summarizes into a matrix; `melt()` converts columns back to rows.
- Load and save CSV files with `pd.read_csv()` and `to_csv(index=False)`.

```python
# Inspecting
print("shape:", sales.shape)
print("columns:", sales.columns.tolist())
print("types:\n", sales.dtypes)
display(sales.head(3), sales.describe(include="all"))

# Selecting and filtering
display(sales.loc[0:2, ["customer", "product", "revenue"]])
display(sales.iloc[:3, :4])
display(sales.loc[sales["city"].isin(["Pune", "Delhi"]) & sales["units"].between(2, 5)])
display(sales.query("unit_price >= 15 and units >= 3"))
print("one cell:", sales.at[0, "customer"], " / positional:", sales.iat[0, 2])

# Sorting, renaming, and making new columns
display(sales.sort_values(["revenue", "customer"], ascending=[False, True]))
display(sales.sort_index(ascending=False).head(3))
display(sales.rename(columns={"unit_price": "price"}).head(2))
display(sales.drop(columns=["discount"]).reset_index(drop=True).head(2))
display(sales.assign(high_value=sales["revenue"] >= 200, city_upper=sales["city"].str.upper()).head())
print("units as strings:\n", sales["units"].astype("string").head())
print("customer initials:\n", sales["customer"].map(lambda name: name[0]).tolist())
print("uppercase products:\n", sales["product"].apply(str.upper).head())
print("discount labels:\n", sales["discount"].fillna(0).replace({0.0: "no discount"}).head())

# Missing values and duplicate rows
print("missing by column:\n", sales.isna().sum())
display(sales.loc[sales["discount"].notna()])
display(sales.fillna({"discount": 0}).head())
print("duplicate flags:", sales.duplicated().tolist())
display(sales.drop_duplicates())
display(sales.dropna(subset=["discount"]))

# Summaries and grouping
print("revenue mean/sum:", sales["revenue"].mean(), sales["revenue"].sum())
print("products:", sales["product"].unique(), "counts:\n", sales["product"].value_counts())
display(sales.groupby("city").agg(order_count=("order_id", "count"), units_sold=("units", "sum"), revenue=("revenue", "sum")))
sales["city_average"] = sales.groupby("city")["revenue"].transform("mean")
display(sales[["city", "revenue", "city_average"]].head())

# Combining tables: concat stacks rows; merge joins on matching keys.
extra = pd.DataFrame({"order_id": [1007], "customer": ["Eli"]})
display(pd.concat([sales[["order_id", "customer"]], extra], ignore_index=True))
city_info = pd.DataFrame({"city": ["Pune", "Mumbai", "Delhi"], "region": ["West", "West", "North"]})
display(sales.merge(city_info, on="city", how="left").head())

# Text and dates
print("city names:", sales["city"].str.lower().unique())
print("Notebook matches:", sales["product"].str.contains("note", case=False).sum())
sales["date"] = pd.to_datetime(sales["date"])
display(sales.assign(month=sales["date"].dt.month, weekday=sales["date"].dt.day_name())[["date", "month", "weekday"]].head())

# Reshape: pivot_table aggregates; melt changes wide data to long data.
display(sales.pivot_table(index="city", columns="product", values="revenue", aggfunc="sum", fill_value=0))
wide = pd.DataFrame({"student": ["Asha", "Ben"], "math": [90, 82], "science": [88, 91]})
display(wide.melt(id_vars="student", var_name="subject", value_name="score"))

# CSV examples (uncomment to use a real file):
# imported = pd.read_csv("sales.csv", parse_dates=["date"])
# imported.to_csv("clean_sales.csv", index=False)
```

```python
# 1-4: preview, select, filter, and sort
print(sales.head(4), "\nshape:", sales.shape)
display(sales[["customer", "product", "revenue"]])
display(sales.loc[(sales["city"] == "Pune") & (sales["units"] >= 4)])
display(sales.nlargest(3, "revenue"))

# 5-8: derive, clean, and summarize
practice = sales.copy()
practice["total_before_discount"] = practice["units"] * practice["unit_price"]
practice["discount_amount"] = practice["total_before_discount"] - practice["revenue"]
print("missing:\n", practice.isna().sum())
practice["discount"] = practice["discount"].fillna(0)
practice = practice.drop_duplicates()
print("rows after removing duplicates:", len(practice))
print("average revenue:", practice["revenue"].mean())
print("total units:", practice["units"].sum())
print("distinct customers:", practice["customer"].nunique())

# 9-11: group, transform, and join
display(practice.groupby("product").agg(revenue=("revenue", "sum"), units=("units", "sum")).sort_values("revenue", ascending=False))
practice["city_average_revenue"] = practice.groupby("city")["revenue"].transform("mean")
city_info = pd.DataFrame({"city": ["Pune", "Mumbai", "Delhi"], "region": ["West", "West", "North"]})
display(practice.merge(city_info, on="city", how="left"))

# 12-15: text, date, pivot, and CSV
display(practice.loc[practice["product"].str.contains("pen", case=False, na=False)])
practice["date"] = pd.to_datetime(practice["date"])
print(practice.groupby(practice["date"].dt.month).size().rename("orders_by_month"))
display(practice.pivot_table(index="city", columns="product", values="revenue", aggfunc="sum", fill_value=0))
# Optional: practice.to_csv("practice_sales.csv", index=False)
```
