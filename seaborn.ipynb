```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["AI", "AI", "CSE", "CSE", "ECE", "ECE"],
    "Marks": [85, 90, 75, 80, 88, 92],
    "Hours": [4, 5, 3, 4, 5, 6]
})

print(df)

sns.set_theme(style="whitegrid")

# Compare average marks by department
sns.barplot(data=df, x="Department", y="Marks")
plt.title("Average Marks by Department")
plt.show()

# Explore the relationship between hours and marks
sns.scatterplot(
    data=df, x="Hours", y="Marks", hue="Department"
)
plt.title("Study Hours vs Marks")
plt.show()

# Compare the distribution of marks
sns.boxplot(data=df, x="Department", y="Marks")
plt.title("Marks Distribution by Department")
plt.show()
```

```python
sns.scatterplot(data=df, x="Hours", y="Marks")
plt.grid(True)
sns.regplot(
    data=df,
    x="Hours",
    y="Marks",
    scatter=False,
    color="red"
)

plt.show()
```
