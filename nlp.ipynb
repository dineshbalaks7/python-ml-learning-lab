```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize,TreebankWordTokenizer


text = "Natural Language Processing (NLP) is a fascinating field's of study. It involves the interaction between computers and humans through natural language. The ultimate objective of NLP is to enable computers to understand, interpret, and generate human language in a valuable way."

# Sentence tokenization
sentences = sent_tokenize(text)
print(sentences)

# Word tokenization
words = word_tokenize(text)
print(words)

# Word punctuation tokenization
wordpuncts = wordpunct_tokenize(text)
print(wordpuncts)

#Treebank Word Tokenization
tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer
# Stemming
porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
snowball_stemmer = SnowballStemmer("english")
```

```python
word = "running, runner, runs, easily, fairer"
words1 = word_tokenize(word)
print(words1)
for w in words1:
    print(w, "->", porter_stemmer.stem(w))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love this phone",
    "I love this camera",
    "This phone is excellent"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF values:")
print(X.toarray().round(3))
```
