```python
import numpy as np

# Create an array
marks = np.array([45, 67, 89, 32, 95, 76, 58])

# Filter marks greater than 70
filtered_marks = marks[marks > 70]
print("Marks above 70:", filtered_marks)

# Calculate mean
print("Average marks:", np.mean(marks))

# Find maximum
print("Maximum marks:", np.max(marks))

# Find minimum
print("Minimum marks:", np.min(marks))
```
