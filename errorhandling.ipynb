```python
try:
    a=int(input())
    b=int(input())
    print(a/b)
except ValueError:
    print("Value Error")
except ZeroDivisionError:
    print("Cannot divide by zero")
except Exception:
    print("Error")
finally:
    print("Program completed")
```

```python
try:
    age=int(input())
    print(age)
except ValueError:
    print("Invalid age !")
```

```python
try:
    a=int(input())
    b=input()
    print(a+b)
except TypeError:
    print("Cannot add integer and string")
```

```python
class AgeError(Exception):
    pass
try:
    age=int(input("Enter your Age: "))
    if(age<18):
        raise AgeError("Age must be 18 or above !")
    print("Eligible")
except AgeError as e:
    print(e)
```

```python
#Zero devision Error
try:
    a=int(input())
    b=int(input())
    print(a/b)
except ZeroDivisionError:
    print("Denominator should not be 0")
```

```python
#Index Error
try:
    numbers = [10, 20, 30, 40, 50]
    n=int(input())
    print(numbers[n])
except IndexError:
    print("Index does not exist")
```

```python
#Key Error
try:
    student = {
    "name": "Dinesh",
    "age": 21,
    "department": "AI&DS"
    }
    key=input()
    print(student[key])
except KeyError:
    print("The Key is not there !")
```
