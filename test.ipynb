```python
#two sum
nums = [1, 2, 3, 4, 5, 6, 7]
target=9
left = 0
right = len(nums) - 1

while left < right:
    total = nums[left] + nums[right]

    if total == target:
        print([nums[left], nums[right]])
        break
    elif total < target:
        left += 1
    else:
        right -= 1
```

```python
#Palindrome using Two Pointers
nums = [1, 2, 3, 2, 1]
right=len(nums)-1
is_palindrome=True
left=0
while left<right:
    if nums[left]!=nums[right]:
        is_palindrome=False
        break
    right-=1
    left+=1
print(is_palindrome)
```

```python
#minimum length of a contiguous subarray whose sum is greater than or equal to target.
nums = [2, 3, 1, 2, 4, 3]
target = 7
left=0
current_sum=0
min_length=float('inf')
for right in range(len(nums)):
    current_sum+=right
    while current_sum>=target:
        min_length=min(min_length,right-left+1)
        current_sum-=nums[left]
        left+=1
print(min_length)
```

```python
#max length of a contiguous subarray whose sum is greater than or equal to target.
nums = [2, 3, 1, 2, 4, 3]
target = 7
left=0
current_sum=0
max_length=float('-inf')
for right in range(len(nums)):
    current_sum+=nums[right]
    while current_sum>=target:
        max_length=max(max_length,right-left+1)
        current_sum-=nums[left]
        left+=1
print(max_length)
```

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
def sum_of_numbers(n):
    if n==0:
        return n
    last=n%10
    return last+sum_of_numbers(n//10)
print(sum_of_numbers(111))
```
