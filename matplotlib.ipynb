```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.grid(True)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Trend")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

plt.figure(figsize=(8, 5))
sns.barplot(x=months, y=sales, palette="Blues_d")
plt.hist(
    months,bins=3,
    color="blue")

plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
